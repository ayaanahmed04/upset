```python
numeric_stat_columns = [
    "SLpM",
    "SApM",
    "TD_Avg",
    "Sub_Avg",
]

percent_stat_columns = [
    "Str_Acc",
    "Str_Def",
    "TD_Acc",
    "TD_Def",
]

zero_counts = {}

for column in numeric_stat_columns:
    zero_counts[column] = (fighters[column] == 0).sum()

for column in percent_stat_columns:
    zero_counts[column] = (fighters[column] == "0%").sum()

zero_audit = pd.DataFrame.from_dict(
    zero_counts,
    orient="index",
    columns=["zero_count"],
)

zero_audit["zero_percent"] = (
    zero_audit["zero_count"] / len(fighters) * 100
)

zero_audit.sort_values("zero_percent", ascending=False)
```

```python
fight_audit = fights.copy()

fight_audit["Event_Date"] = pd.to_datetime(
    fight_audit["Event_Date"]
)

fight_audit["Year"] = fight_audit["Event_Date"].dt.year

core_stat_columns = [
    "F1_KD",
    "F2_KD",
    "F1_Sig_Landed",
    "F1_Sig_Att",
    "F2_Sig_Landed",
    "F2_Sig_Att",
    "F1_TD_Landed",
    "F2_TD_Landed",
    "F1_TD_Att",
    "F2_TD_Att",
    "F1_Sub_Att",
    "F2_Sub_Att",
    "F1_Ctrl_Sec",
    "F2_Ctrl_Sec",
]

fight_audit["all_core_stats_zero"] = (
    fight_audit[core_stat_columns] == 0
).all(axis=1)

zero_stats_by_year = (
    fight_audit
    .groupby("Year")
    .agg(
        total_fights=("Fight_URL", "count"),
        all_zero_fights=("all_core_stats_zero", "sum"),
    )
)

zero_stats_by_year["zero_percent"] = (
    zero_stats_by_year["all_zero_fights"]
    / zero_stats_by_year["total_fights"]
    * 100
)

zero_stats_by_year
```

```python
fight_audit["no_sig_strike_stats"] = (
    (fight_audit["F1_Sig_Att"] == 0)
    & (fight_audit["F2_Sig_Att"] == 0)
)

fight_audit["no_takedown_stats"] = (
    (fight_audit["F1_TD_Att"] == 0)
    & (fight_audit["F2_TD_Att"] == 0)
)

fight_audit["no_control_stats"] = (
    (fight_audit["F1_Ctrl_Sec"] == 0)
    & (fight_audit["F2_Ctrl_Sec"] == 0)
)

stats_by_year = (
    fight_audit
    .groupby("Year")
    .agg(
        total_fights=("Fight_URL", "count"),
        no_sig_strike_stats=("no_sig_strike_stats", "sum"),
        no_takedown_stats=("no_takedown_stats", "sum"),
        no_control_stats=("no_control_stats", "sum"),
    )
)

stats_by_year["no_sig_strike_percent"] = (
    stats_by_year["no_sig_strike_stats"]
    / stats_by_year["total_fights"]
    * 100
)

stats_by_year["no_takedown_percent"] = (
    stats_by_year["no_takedown_stats"]
    / stats_by_year["total_fights"]
    * 100
)

stats_by_year["no_control_percent"] = (
    stats_by_year["no_control_stats"]
    / stats_by_year["total_fights"]
    * 100
)

stats_by_year
```

```python
fight_audit["has_any_control"] = (
    (fight_audit["F1_Ctrl_Sec"] > 0)
    | (fight_audit["F2_Ctrl_Sec"] > 0)
)

control_by_event = (
    fight_audit
    .groupby("Event_Date")
    .agg(
        total_fights=("Fight_URL", "count"),
        fights_with_control=("has_any_control", "sum"),
    )
    .reset_index()
)

control_by_event["control_signal_percent"] = (
    control_by_event["fights_with_control"]
    / control_by_event["total_fights"]
    * 100
)
```

### Provisional finding: control-time coverage

Control-time values appear to have a structural coverage boundary around
July 16, 1999 (UFC 21). Earlier fights frequently contain zeros for both
fighters, which may represent unavailable data rather than true zero control.
This requires source validation before cleaning.

**Evidence:** All 7 fights from UFC 20 (1999-05-07) have zero control
time for both fighters, while all 8 fights from the immediately following
UFC 21 (1999-07-16) contain a positive control-time value for at least one
fighter.

**Working assumption:** Pre-UFC-21 control-time zeros represent unavailable
data rather than true zero control. Do not modify raw data; handle this later
during normalization/cleaning.

## Profile-Only Fighter Investigation

The fighter profile dataset contains 1,801 names that do not appear in the historical fight dataset.

The goal of this section is not to delete these fighters. We want to briefly investigate why the profile dataset has broader coverage than the fight dataset and determine what role these records should have in UPSET.

```python
# Collect every fighter name that appears in the historical fight table.
fight_fighter_names = set(fights["Fighter_1"].dropna()) | set(
    fights["Fighter_2"].dropna()
)

# Select fighter profiles whose names never appear in either fight column.
profile_only = fighters[
    ~fighters["Fighter_Name"].isin(fight_fighter_names)
].copy()

print(f"Profile-only fighter rows: {len(profile_only):,}")
print()

profile_only[
    [
        "Fighter_Name",
        "Weight",
        "Wins",
        "Losses",
        "Draws",
        "Fighter_URL",
    ]
].head(25)
```

### Profile-Only Fighter Conclusion

The fighter profile dataset contains 1,803 profile-only rows representing 1,801 unique names.

The row/name difference is caused by two duplicate-name identity pairs:
- Tony Johnson
- Victor Valenzuela

All profile-only records use `ufcstats.com` fighter URLs.

External validation suggests that UFCStats' fighter directory has broader scope than the historical UFC fight dataset. Some profileonly fighters have records from UFC adjacent or other promotions such as DWCS, Strikeforce, and ProElite.

Therefore, profile-only fighters should not be treated as invalid records or deleted.

Working conclusion:

- the fighter profile table has broader coverage than the fight table
- the fight table should define the historical UFC fight universe for this dataset
- fighter profiles may still be useful for identity resolution, enrichment, and future analysis
- filtering should happen for a specific analytical purpose rather than by deleting profile-only fighters globally

```python
round_stats_df = pd.DataFrame(rounds_data["data"])

round_stats_df[
    [
        "fighterName",
        "round",
        "knockdowns",
        "significantStrikes",
        "totalStrikes",
        "takedowns",
        "submissionAttempts",
        "controlTime",
        "head",
        "body",
        "leg",
        "distance",
        "clinch",
        "ground",
    ]
]
```

```python
def parse_landed_attempted(value):
    landed, attempted = value.split(" of ")
    return int(landed), int(attempted)


validation_rows = []

for _, row in round_stats_df.iterrows():
    sig_landed, sig_attempted = parse_landed_attempted(
        row["significantStrikes"]
    )

    head_landed, head_attempted = parse_landed_attempted(row["head"])
    body_landed, body_attempted = parse_landed_attempted(row["body"])
    leg_landed, leg_attempted = parse_landed_attempted(row["leg"])

    distance_landed, distance_attempted = parse_landed_attempted(
        row["distance"]
    )
    clinch_landed, clinch_attempted = parse_landed_attempted(row["clinch"])
    ground_landed, ground_attempted = parse_landed_attempted(row["ground"])

    validation_rows.append({
        "fighter": row["fighterName"],
        "round": row["round"],
        "target_landed_matches": (
            head_landed + body_landed + leg_landed == sig_landed
        ),
        "target_attempted_matches": (
            head_attempted + body_attempted + leg_attempted == sig_attempted
        ),
        "position_landed_matches": (
            distance_landed + clinch_landed + ground_landed == sig_landed
        ),
        "position_attempted_matches": (
            distance_attempted
            + clinch_attempted
            + ground_attempted
            == sig_attempted
        ),
    })

validation_df = pd.DataFrame(validation_rows)
validation_df
```
